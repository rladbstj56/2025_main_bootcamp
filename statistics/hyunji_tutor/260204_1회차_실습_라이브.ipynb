### 1. 양적 데이터 (Quantitative Data)

숫자로 측정할 수 있으며, **크기·차이·연산**이 의미 있는 데이터

| 구분 | 설명 | 예시 |
|---|---|---|
| 연속형 (Continuous) | 측정 가능한 실수 값 | 키, 몸무게, 온도, 시간 |
| 이산형 (Discrete) | 측정 가능한 정수 값 | 구매 횟수, 방문자 수, 자녀 수 |

### 2. 질적 데이터 (Qualitative Data)

| 구분 | 설명 | 예시 |
|---|---|---|
| 명목형 (Nominal) | 순서가 없는 범주 | 성별, 혈액형, 국가 |
| 순서형 (Ordinal) | 순서는 있으나 간격의 크기는 의미 없음 | 만족도, 선호도 순위, 등급 |

### 실습5. 신뢰구간 구하기

```python
sample_data = sample_df['토지면적']

#표본 개수 
n = len(sample_data)
#표본의 평균
sample_mean = np.mean(sample_data)
#표본의 표준편차
s = np.std(sample_data,ddof=1)
#표준오차의 근사값
se = s/np.sqrt(n)

# 신뢰구간 구하기
stats.t.interval(confidence=0.95, df = n-1, loc = sample_mean , scale=se)
```

```python
# 신뢰구간 구하기
stats.t.interval(confidence=0.2, df = n-1, loc = sample_mean , scale=se)
```

### 추가적으로, 
### 100번의 독립적인 실험을 통해 100개의 신뢰구간을 구한다면 어떤 일이 벌어지는지 

```python
import numpy as np
import scipy.stats as st

# 설정
pop_mean = 50       # 실제 모평균 (우리는 알고 있다고 가정)
pop_std = 10        # 모표준편차
sample_size = 30
conf_level = 0.95
iterations = 100

success_count = 0

for _ in range(iterations):
    # 1. 샘플 추출
    sample = np.random.normal(pop_mean, pop_std, sample_size)
    
    # 2. 신뢰구간 계산
    mean = np.mean(sample)
    se = pop_std / np.sqrt(sample_size)
    ci = st.norm.interval(conf_level, loc=mean, scale=se)
    
    # 3. 모평균 포함 여부 확인
    if ci[0] <= pop_mean <= ci[1]:
        success_count += 1

print(f"100개의 구간 중 모평균을 포함한 구간의 개수: {success_count}")
```
