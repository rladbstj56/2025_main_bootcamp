loc vs iloc
- loc: 라벨 기준으로 행을 불러오는 것
    - 라벨로 찾기 
    - 슬라이싱 하려면 ["monday":"wednesday"] 와 같이 적을 것
    - 행 지정하려면 [["monday","wednesday"]]
- iloc: 인덱스 몇번째 행인지를 기준으로 불러오는 것
    - 순서로 찾기
    - 슬라이싱 하려면 [4:6]
    - 행 지정하려면 [[2,4,6]]

```python
titanic.iloc[[0, 4, 5]] # bracket 2개 필요
# iloc 는 순서 기반
```

```python
titanic.loc[[0, 4, 5]] 
# 우연히 인덱스값과 라벨이 같은 것 뿐
```

```python
titanic.iloc[5:9]
```
